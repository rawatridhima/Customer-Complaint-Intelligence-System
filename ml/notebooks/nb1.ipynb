```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/complaints.csv",
    # usecols=cols,
    dtype=str,
    nrows=500_000
)
df.head(5)
```

```python
TEXT = "Consumer complaint narrative"

# 1. How many rows actually have text?
print(df[TEXT].notna().sum())

# 2. Date range of this snapshot (write it in your README)
dates = pd.to_datetime(df["Date received"], errors="coerce")
print(dates.min(), dates.max())

# 3. Keep only rows with text, then look at the labels
df = df.dropna(subset=[TEXT])
print(df.shape)
print(df["Product"].value_counts())
print(df["Issue"].value_counts().head(40))

# 4. Read a couple of real complaints
for t in df[TEXT].sample(3, random_state=42):
    print(t[:400], "\n---")
```
